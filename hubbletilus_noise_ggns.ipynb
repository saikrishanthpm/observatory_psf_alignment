# Hubble with noise: GGNS instead of Nautilus

A clone of `hubbletilus_noise.ipynb` in which the nested sampler is GGNS (gradient-guided nested sampling, `gradNS.HamiltonianNS`; Lemos et al. 2024, ICML, https://github.com/Pablo-Lemos/GGNS). GGNS moves live points with Hamiltonian slice sampling, which uses the gradient of the log-likelihood, so the likelihood here is evaluated with Optiland's PyTorch backend and differentiated by autograd. Everything else (covariances, observations, differential evolution, OPD checks) is unchanged and runs with the numpy backend.

**Changes from `hubbletilus_noise.ipynb`**
1. **Sampler.** Nautilus is replaced by `HamiltonianNS`. The likelihood is a batched torch version of `loglike` (Section "GGNS").
2. **Optiland gradient fix.** Optiland 0.6.0 converts the reference-sphere center and radius to Python floats (`.item()` in `ChiefRayStrategy._create_spherical_ref`), which makes the decenter and tilt gradients wrong. That method is replaced at runtime by one that keeps them as tensors (as in `hubble_gradfit.ipynb`); the installed package is not modified.
3. **`hubble()`** converts the tilts with `* np.pi / 180` instead of `np.deg2rad`, which does not accept tensors that require gradients. The numbers are the same.
4. **Prior.** GGNS samples the unit cube, and the likelihood maps it to the same uniform box as `ptform` ($\pm200\,\mu$m, $\pm0.167^\circ$), **without the rounding to two decimals**, which has no gradient. Sampling in the unit cube also keeps micrometers and degrees on comparable scales for the Hamiltonian steps.
5. **Normalization.** `lnorm` now uses the 15 retained modes instead of 5, so $\ln Z$ comes out correctly normalized (the thesis corrected the Nautilus value by $-45.95$).
6. **Noise** in the synthetic observations is drawn with a fixed seed, so a run can be repeated.
7. **DE start and bounds** use the 100 highest-likelihood GGNS samples, the counterpart of the final 100 Nautilus samples (its innermost shell); GGNS returns the dead points sorted by likelihood.
8. **DE log file** is `solutions_log_noise_ggns.txt`, so the thesis log `solutions_log_noise.txt` is not appended to.

**Before running**
- Install GGNS and the two packages it imports at load time into this environment (`spaceklip_new`); none of them is installed yet:
  `pip install getdist anesthetic git+https://github.com/Pablo-Lemos/GGNS.git`
- GGNS works in float32 (a module-level constant); the likelihood converts to float64 internally and back.
- Cost: every point at which GGNS needs a gradient costs one forward and one backward pass through five PSFs, about 0.2 s on this Mac's CPU, compared with 0.05 s for a numpy likelihood call. The number of evaluations GGNS needs here is not known in advance; run the cost check in the GGNS section first and watch the progress output.

```python
def _is_odd(integer):

    return integer & 1

def noll_indices(j):

    if j < 1:
        raise ValueError("Zernike index j must be a positive integer.")

    # from i, compute m and n
    # I'm not sure if there is an easier/cleaner algorithm or not.
    # This seems semi-complicated to me...

    # figure out which row of the triangle we're in (easy):
    n = int(np.ceil((-1 + np.sqrt(1 + 8 * j)) / 2) - 1)
    if n == 0:
        m = 0
    else:
        nprev = (n + 1) * (n + 2) / 2  # figure out which entry in the row (harder)
        # The rule is that the even Z obtain even indices j, the odd Z odd indices j.
        # Within a given n, lower values of m obtain lower j.

        resid = int(j - nprev - 1)

        if _is_odd(j):
            sign = -1
        else:
            sign = 1

        if _is_odd(n):
            row_m = [1, 1]
        else:
            row_m = [0]

        for i in range(int(np.floor(n / 2.))):
            row_m.append(row_m[-1] + 2)
            row_m.append(row_m[-1])

        m = row_m[resid] * sign

    return n, m

def circ_bessel(n, m, npix=128, rho=None, theta=None, outside=np.nan, zero=19.053609871416903, **kwargs):
 
    if theta is None and rho is None:
        x = (np.arange(npix, dtype=np.float64) - (npix - 1) / 2.) / (zero)
        y = x
        xx, yy = np.meshgrid(x, y)

        rho = np.sqrt(xx ** 2 + yy ** 2)
        theta = np.arctan2(yy, xx)
    elif (theta is None and rho is not None) or (theta is not None and rho is None):
        raise ValueError("If you provide either the `theta` or `rho` input array, you must "
                         "provide both of them.")
        
    if not np.all(np.asarray(rho.shape == theta.shape)):
        raise ValueError('The rho and theta arrays do not have consistent shape.')

    if m == 0:
        if n == 0:
            cb_result = j1(2*np.pi*rho) * (1/rho)
        else:
            cb_result = -1**(n/2 + np.abs(m)) * jv(n+1, 2*np.pi*rho) * (1/rho)
    elif m > 0:
        cb_result = -1**(n/2 + np.abs(m)) * jv(n+1, 2*np.pi*rho) * (1/rho) * np.cos(np.abs(m) * theta)
    else:
        cb_result = -1**(n/2 + np.abs(m)) * jv(n+1, 2*np.pi*rho) * (1/rho) * np.sin(np.abs(m) * theta)
    return cb_result


def circ_bessel1(j, **kwargs):

    n, m = noll_indices(j)
    return circ_bessel(n, m, **kwargs)

def basis_vectors(nterms=15, npix=256, rho=None, theta=None, **kwargs):
    
    if rho is not None and theta is not None:
        # both are required, but validated in circbessel1
        shape = rho.shape
        use_polar = True
    elif (theta is None and rho is not None) or (theta is not None and rho is None):
        raise ValueError("If you provide either the `theta` or `rho` input array, you must "
                         "provide both of them.")

    else:
        shape = (npix, npix)
        use_polar = False

    basis_output = np.zeros((nterms,) + shape)

    if use_polar:
        for j in range(nterms):
            basis_output[j] = circ_bessel1(j+1, rho=rho, theta=theta, **kwargs)
    else:
        for j in range(nterms):
            basis_output[j] = circ_bessel1(j+1, npix=npix, **kwargs)
    return basis_output
```

```python
from functools import lru_cache

# --- drop-in helpers (uses your existing basis_vectors) ---

def _outside_key(outside):
    if outside is None or (isinstance(outside, float) and np.isnan(outside)):
        return "nan"
    return float(outside)

@lru_cache(maxsize=16)
def _polar_grid(npix, zero):
    x = (np.arange(npix, dtype=np.float64) - (npix - 1) / 2.0) / zero
    xx, yy = np.meshgrid(x, x)
    rho = np.sqrt(xx ** 2 + yy ** 2)
    theta = np.arctan2(yy, xx)
    return rho, theta

@lru_cache(maxsize=16)
def basis_vectors_cached(nterms=15, npix=256, zero=19.053609871416903, outside="nan"):
    """Cached wrapper around basis_vectors. `outside` must be hashable ('nan' or a float)."""
    outside_val = np.nan if outside == "nan" else float(outside)
    rho, theta = _polar_grid(npix, zero)
    return basis_vectors(
        nterms=nterms,
        rho=rho,
        theta=theta,
        outside=outside_val,
        zero=zero,
    )

def clear_basis_cache():
    basis_vectors_cached.cache_clear()
    _polar_grid.cache_clear()

def decompose_psf_cached(psf, nterms=15, outside=np.nan, **kwargs):
    npix = psf.shape[0]
    zero = kwargs.get("zero", 19.053609871416903)
    basis_stack = basis_vectors_cached(
        nterms=nterms,
        npix=npix,
        zero=zero,
        outside=_outside_key(outside),
    )
    basis_flat = basis_stack.reshape(nterms, -1)
    psf_flat = psf.ravel()
    numerators = basis_flat @ psf_flat
    denominators = np.einsum("ij,ij->i", basis_flat, basis_flat)
    return numerators / denominators
```

```python
def decompose_psf(psf, nterms=10, basis=zernike_basis, **kwargs):
    npix = psf.shape[0]
    basis_set = basis(nterms=nterms, npix=npix, outside=0.0, **kwargs)
    basis_set = np.array(basis_set)
    basis_flat = basis_set.reshape(nterms, -1)
    numerators = basis_flat @ psf.ravel()
    denominators = np.einsum('ij,ij->i', basis_flat, basis_flat)
    coeffs = numerators / denominators
    return coeffs
```

```python
def hubble(WL=0.55,sdx=0,sdy=0,sdz=0,stx=0,sty=0):

    K=1000 #mm to micron
    #Hubble
    hubble = optic.Optic(name="Hubble")

    hubble.add_surface(index=0, radius=np.inf, thickness=np.inf)

    # add physical aperture in primary mirror
    obscuration = physical_apertures.RadialAperture(r_max=1200, r_min=384)
    c1 = physical_apertures.OffsetRadialAperture(r_max=75, r_min=0, offset_x=0, offset_y=1000)
    c2 = physical_apertures.OffsetRadialAperture(r_max=75, r_min=0, offset_x=-707.15, offset_y=-707.15)
    c3 = physical_apertures.OffsetRadialAperture(r_max=75, r_min=0, offset_x=707.15, offset_y=-707.15)
    r1 = physical_apertures.RectangularAperture(x_min=-19, x_max=19, y_min=-1200, y_max=1200)
    r2 = physical_apertures.RectangularAperture(x_min=-1200, x_max=1200, y_min=-19, y_max=19)

    ap = obscuration - r1 - r2 -c1 - c2 - c3

    hubble.add_surface(
        index=1,
        x=0.0, y=0.0, z=0,   # paraxial / entrance reference
    )

    hubble.add_surface(
        index=2,
        radius=-11040.02286,
        x=0.0, y=0.0, z=4910.01016,
        material="mirror",
        is_stop=True,
        conic=-1.001152,
        aperture=ap,
    )

    hubble.add_surface(
        index=3,
        radius=-1349.31166,
        x=sdx / K,
        y=sdy / K,
        z=0 + sdz / K,              # independent M2 piston
        rx=stx * np.pi / 180,       # np.deg2rad rejects tensors that require grad
        ry=sty * np.pi / 180,
        material="mirror",
        conic=-1.483014,
    )

    hubble.add_surface(
        index=4,
        radius=-635.38227,
        x=0.0, y=0.0, z=6365.20955,         # image fixed while M2 moves
    )

    hubble.set_aperture(aperture_type="EPD", value=2400)

    hubble.set_field_type(field_type="angle")
    hubble.add_field(y=0.)
    hubble.add_field(y=0.038)
    hubble.add_field(y=-0.038)

    hubble.add_wavelength(value=WL, is_primary=True)
    
    return hubble
```

```python
def genpsf(WL=0.55, sdx=0, sdy=0, sdz=0, stx=0, sty=0, dterms=5, field =(0,0)):

    telescope = hubble(WL, sdx, sdy, sdz, stx, sty)
    psf = MMDFTPSF(telescope, field=field, wavelength=WL, grid_size=64)
    PSF = psf.psf[96:160,96:160]
    r = decompose_psf_cached(PSF, nterms=dterms)

    return r
```

```python
def genpsfn(WL=0.55, sdx=0, sdy=0, sdz=0, stx=0, sty=0, dterms=5, field =(0,0)):

    telescope = hubble(WL, sdx, sdy, sdz, stx, sty)
    psf = MMDFTPSF(telescope, field=field, wavelength=WL, grid_size=64)
    PSF = psf.psf[96:160,96:160]
    gauss_noise = np.random.normal(0, 0.01, PSF.shape)
    r = decompose_psf_cached(PSF + gauss_noise, nterms=dterms)

    return r
```

## GGNS

The likelihood is the same as `loglike`: the five fields with their covariances and normalizations, written with torch operations. GGNS calls it with a batch of points of shape `(n, 5)` in the unit cube, sometimes with `requires_grad=True` (to take the gradient with autograd) and sometimes without. `_LogLike` computes the value and the gradient one point at a time and hands the stored gradient to autograd, so memory does not grow with the batch size.

```python
_OPTILAND_REF = strategy.ChiefRayStrategy._create_spherical_ref


def _create_spherical_ref(self, x, y, z):
    # Optiland's version without .item(), so the reference sphere stays in the autodiff graph
    R = be.sqrt(x**2 + y**2 + (z - self.pupil_z) ** 2)
    return SphericalReference((x.reshape(()), y.reshape(()), z.reshape(())), R.reshape(()))


def use_numpy():
    be.set_backend("numpy")
    strategy.ChiefRayStrategy._create_spherical_ref = _OPTILAND_REF


def use_torch():
    be.set_backend("torch")
    be.set_precision("float64")
    be.grad_mode.enable()
    strategy.ChiefRayStrategy._create_spherical_ref = _create_spherical_ref


F64 = torch.float64
FIELDS = [(0, 0), (0, 1), (1, 0), (1, 1), (-1, -1)]
HALF_NP = np.array([200.0, 200.0, 200.0, 0.167, 0.167])   # half-widths of the uniform prior box (ptform)
HALF = torch.tensor(HALF_NP, dtype=F64)

_basis = basis_vectors_cached(nterms=15, npix=64, zero=19.053609871416903, outside="nan").reshape(15, -1)
BASIS_T = torch.tensor(_basis, dtype=F64)
NORM_T = torch.tensor(np.einsum("ij,ij->i", _basis, _basis), dtype=F64)
OBS_T = torch.tensor(np.array([ct, ctf01, ctf10, ctf11, ctfm11]), dtype=F64)
CINV_T = torch.tensor(np.array([Cinv, Cf01inv, Cf10inv, Cf11inv, Cfm11inv]), dtype=F64)
LNORM_T = torch.tensor(float(lnorm + lnormf01 + lnormf10 + lnormf11 + lnormfm11), dtype=F64)


def loglike_one(u):
    # loglike for one point u in the unit cube (float64 tensor of shape (5,)); no rounding
    theta = HALF * (2 * u - 1)
    total = LNORM_T
    for field, obs, cinv in zip(FIELDS, OBS_T, CINV_T):
        psf = MMDFTPSF(hubble(0.55, *theta), field=field, wavelength=0.55, grid_size=64).psf[96:160, 96:160]
        r = (BASIS_T @ psf.reshape(-1)) / NORM_T - obs
        total = total - 0.5 * r @ cinv @ r
    return total


class _LogLike(torch.autograd.Function):
    # value and gradient one point at a time; backward returns the stored gradients

    @staticmethod
    def forward(ctx, u):
        values, grads = [], []
        for row in u:
            with torch.enable_grad():
                x = row.detach().clone().requires_grad_(True)
                v = loglike_one(x)
                (g,) = torch.autograd.grad(v, x)
            values.append(v.detach())
            grads.append(g)
        ctx.save_for_backward(torch.stack(grads))
        return torch.stack(values)

    @staticmethod
    def backward(ctx, grad_out):
        (grads,) = ctx.saved_tensors
        return grad_out[:, None] * grads


def loglike_ggns(u):
    # GGNS likelihood: float32 tensor (n, 5) in the unit cube -> float32 tensor (n,)
    assert be.get_backend() == "torch", "call use_torch() first"
    u64 = u.to(F64)
    if u.requires_grad:
        out = _LogLike.apply(u64)
    else:
        with torch.no_grad():
            out = torch.stack([loglike_one(row) for row in u64])
    return out.to(u.dtype)
```

Checks before the run: the torch likelihood should match `loglike` (numpy) at the truth, the autograd gradient should match central differences of `loglike`, and the timing gives the cost per point with and without the gradient.

```python
theta_true = np.array([sdx_true, sdy_true, sdz_true, stx_true, sty_true])
theta_test = np.array([40.0, -100.0, 30.0, 0.12, -0.03])
to_u = lambda theta: (np.asarray(theta) / HALF_NP + 1) / 2

use_numpy()
ll_np = loglike(theta_true)
h = 1e-4   # central-difference step in the unit cube
fd = []
for k in range(5):
    e = np.zeros(5)
    e[k] = h
    fd.append((loglike(HALF_NP * (2 * (to_u(theta_test) + e) - 1)) - loglike(HALF_NP * (2 * (to_u(theta_test) - e) - 1))) / (2 * h))

use_torch()
ll_t = loglike_one(torch.tensor(to_u(theta_true), dtype=F64)).item()
u = torch.tensor(to_u(theta_test), dtype=torch.float32)[None, :].requires_grad_(True)
loglike_ggns(u).sum().backward()

print(f"loglike at the truth: numpy {ll_np:.8f}, torch {ll_t:.8f}, difference {ll_t - ll_np:.1e}")
print("d lnL / du at theta_test")
print("  central differences:", np.array(fd))
print("  autograd (float32): ", u.grad[0].numpy())

batch = torch.rand(4, 5)
t0 = time.perf_counter()
loglike_ggns(batch)
t1 = time.perf_counter()
loglike_ggns(batch.clone().requires_grad_(True)).sum().backward()
t2 = time.perf_counter()
print(f"per point: {(t1 - t0) / 4:.2f} s without gradient, {(t2 - t1) / 4:.2f} s with gradient")
```

```python
NLIVE = 100   # evidence error ~ sqrt(H / NLIVE); the GGNS examples use 500 for 5 parameters

params = [
    Param(name=name, prior_type="Uniform", prior=(0.0, 1.0), label=label)
    for name, label in zip(["sdx", "sdy", "sdz", "stx", "sty"], ["dx", "dy", "dz", "tx", "ty"])
]
ns = HamiltonianNS(
    loglike=loglike_ggns,
    params=params,
    nlive=NLIVE,
    tol=1e-2,
    clustering=False,
    verbose=True,
    device=torch.device("cpu"),
)

use_torch()
t0 = time.perf_counter()
ns.run()
print(f"GGNS finished in {(time.perf_counter() - t0) / 3600:.2f} h with {ns.get_like_evals()} likelihood evaluations")
ns.save("hubbletilus_noise_ggns.pkl")
```
