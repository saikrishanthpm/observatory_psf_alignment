```python
from optiland.fileio import load_zemax_file
from optiland.wavefront import OPD
import optiland.backend as be
from optiland.psf import FFTPSF
```

```python
be.set_backend("numpy")
be.set_precision("float64")
```

```python
tel = load_zemax_file("pearl.zmx")
```

```python
OPD(tel, field=(0, 0.17/0.285), wavelength=0.65, num_rays=128).view()
```

```python
OPD(tel, field=(0, 0.5965), wavelength=0.65, num_rays=128).view()
```

```python
OPD(tel, field=(0, 0.1929), wavelength=0.65, num_rays=128).view()
```

```python
FFTPSF(tel, wavelength=0.65, field=(0,0.5965),num_rays=512).view()
```
